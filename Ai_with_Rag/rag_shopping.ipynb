```python
from sklearn.metrics.pairwise import cosine_similarity

products = [
    "Running shoes for exercise",
    "Cotton T-shirt for casual wear",
    "Water bottle for drinking"
]

product_embeddings = model.encode(products)

question = "Shoes for jogging"
question_embedding = model.encode(question)

scores = cosine_similarity(
    [question_embedding],
    product_embeddings
)[0]

score = cosine_similarity(
    [question_embedding],
    product_embeddings
)

print(score,"score------")
print(scores,"scores---------")

for product, score in zip(products, scores):
    print(product, "->", round(float(score), 3))
```

```python

import faiss
import numpy as np

dimension = product_embeddings.shape[1]

index = faiss.IndexFlatL2(dimension)

print("product_embeddings",product_embeddings.shape)
print("Embedding dimension:", dimension)
print("FAISS index created!")
```

```python

vectors = np.array(
    product_embeddings,
    dtype=np.float32
)

index.add(vectors)

print("Number of products stored:", index.ntotal)
```

```python

question = "Shoes for jogging"

question_embedding = model.encode(question)

query_vector = np.array(
    [question_embedding],
    dtype=np.float32
)

distances, indices = index.search(query_vector, 1)

print(products,"products-------")
print(distances,'distances===========')
print("Matched product:", products[indices[0][0]])
print("Distance:", distances[0][0])

print("FAISS index:", indices[0][0])
print("FAISS matched:", products[indices[0][0]])
print("First product:", products[0])
```

```python

question = "Shoes for jogging"

retrieved_product = products[indices[0][0]]

prompt = f"""
Customer question: {question}

Product information: {retrieved_product}

Answer the customer using the product information.
"""

print(prompt)
```

```python

def ask_shopping_question(question):

    # Step 1: Convert question into embedding
    question_embedding = model.encode(question)

    # Step 2: Prepare vector for FAISS
    query_vector = np.array(
        [question_embedding],
        dtype=np.float32
    )

    # Step 3: Retrieve the best matching product
    distances, indices = index.search(query_vector, 1)

    product_index = indices[0][0]
    retrieved_product = products[product_index]

    # Step 4: Create prompt
    prompt = f"""
    You are a shopping assistant.

    Customer question: {question}

    Available product information: {retrieved_product}

    Answer using the product information.
    If the information is insufficient, say so.
    """

    # Step 5: Generate answer using Gemini
    response = client.models.generate_content(
        model="gemini-2.5-flash",
        contents=prompt
    )

    return response.text
```

```python

answer = ask_shopping_question(
    "Which product good for wearing?"
)

print(answer)
```

```python

questions = [
    "Which shoes can I use for jogging?",
    "I need a shirt for casual wear.",
    "What can I use to store drinking water?"
]

for question in questions:
    print("Question:", question)

    answer = ask_shopping_question(question)

    print("Answer:", answer)
    print("-" * 50)
```
